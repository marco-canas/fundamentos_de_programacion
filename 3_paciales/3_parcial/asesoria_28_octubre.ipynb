b) **(1.5 pts)** Escriba una función `criba_vectorizada(N)` que use **NumPy** pero **sin ciclos explícitos de Python** para marcar los múltiplos (puede usar asignación con *slicing*: `es_primo[p*p::p] = False`). De tal manera que tal función entregue la lista de los primos menores que $N$. 

```python
import numpy as np 
def criba_vectorizada(N):
    lista_primos = []
    vector_num = np.array(list(range(2, N + 1)))
    while len(vector_num) > 0: 
        lista_primos.append(vector_num[0])
        vector_num = vector_num[vector_num%vector_num[0]!=0]
    return lista_primos 
    
    
```
