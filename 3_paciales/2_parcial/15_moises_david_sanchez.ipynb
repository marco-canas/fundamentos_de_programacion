### Ejercicio 2.1 — Fibonacci + primalidad con `for` (12 pts)

* Usando un ciclo `for`, genere los primeros `N = 20` términos de la sucesión de Fibonacci.   
* Luego, determine cuáles de esos términos son **primos** e imprima la lista de Fibonacci primos.



````python
f1 = 1
f2 = 1
lista_numeros_fibonacci = [f1, f2]  

for i in range(18):
    f3 = f1+f2
    lista_numeros_fibonacci.append(f3)
    f1 = f2
    f2 = f3
    
lista_numeros_fibonacci    
````

Plan

1. creo una función que identifica primos
2. pongo a recorrer esta función a toda la lista de los términos de fibonacci que tengo y al que identifique que sea primo, lo meta en una nueva lista. 

````python
def es_primo(n):
    cuenta = 0
    for d in range(1, n+1): 
        if n%d == 0: 
            cuenta = cuenta + 1
    if cuenta == 2: 
         return True 
````

````python
es_primo(4)
````

````python
lista_de_primos_de_fibonacci = []
for f in lista_numeros_fibonacci:
    if es_primo(f): 
        lista_de_primos_de_fibonacci.append(f)    
lista_de_primos_de_fibonacci        
````
