**1.** Escriba una función `clasificar_imc(peso, altura)` que retorne:
- `"Bajo peso"` si IMC < 18.5
- `"Normal"` si 18.5 ≤ IMC < 25
- `"Sobrepeso"` si 25 ≤ IMC < 30
- `"Obesidad"` si IMC ≥ 30

Donde IMC = peso / altura².


```python
def clasificar_imc(peso, altura): 
    imc = peso/altura**2
    if imc < 18.5:
        return "Bajo peso"
    elif 18.5<=imc < 25:
        return "Normal"
    elif 25<=imc < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"
```


**17.** Escriba una función `normalizar(lista)` que retorne una nueva lista con cada valor transformado como `(x - min) / (max - min)`. Considere el caso donde `max == min`.


```python
X = [10,15, 23, 51, 20]
def normalizar(lista):
    minimo = min(lista)
    maximo = max(lista)
    nueva_lista = []
    for x in lista: 
        nueva_lista.append((x-minimo)/(maximo-minimo))
    return nueva_lista    
```

```python
normalizar(X)
```

```python

# 16
def media(lista):
    if not lista: return None
    return sum(lista) / len(lista)
```

```python

# 17
def normalizar(lista):
    if not lista: return []
    mn, mx = min(lista), max(lista)
    if mx == mn: return [0.0] * len(lista)
    return [(x - mn) / (mx - mn) for x in lista]
```

```python

# 18
def contar_frecuencias(lista):
    frec = {}
    for x in lista:
        frec[x] = frec.get(x, 0) + 1
    return frec
```

```python

# 19
def filtrar_por_umbral(lista, umbral):
    return [x for x in lista if x > umbral]
```

```python

# 20
def resumen_estadistico(lista):
    if not lista:
        return {"min": None, "max": None, "media": None, "n": 0}
    return {
        "min": min(lista),
        "max": max(lista),
        "media": sum(lista) / len(lista),
        "n": len(lista)
    }
```
