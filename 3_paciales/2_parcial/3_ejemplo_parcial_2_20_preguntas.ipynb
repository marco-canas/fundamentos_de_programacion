````python
# 1
def categoria_riesgo(p):
    if p < 0.2: return "Muy bajo"
    elif p < 0.4: return "Bajo"
    elif p < 0.6: return "Medio"
    elif p < 0.8: return "Alto"
    else: return "Muy alto"

# 2
predicciones = [0,1,1,0,1,0,0,1]
reales = [0,1,0,0,1,1,0,1]
for p, r in zip(predicciones, reales):
    if p == 1 and r == 1: print("TP")
    elif p == 0 and r == 0: print("TN")
    elif p == 1 and r == 0: print("FP")
    else: print("FN")

# 3
def es_mejora(actual, anterior):
    if actual > anterior: return True
    elif actual == anterior: return False
    else: return "empeoró"

# 4
metricas = {"precision": 0.82, "recall": 0.75, "f1": 0.78}
if all(v > 0.75 for v in metricas.values()):
    print("Modelo balanceado")
else:
    for k, v in metricas.items():
        if v <= 0.75:
            print(f"Revisar {k}")

# 5
def validar_dato(x):
    if x is None: return "nulo"
    if isinstance(x, bool): return "inválido"
    if isinstance(x, (int, float)):
        if x != x: return "inválido"   # NaN
        return "numérico válido"
    return "inválido"

# 6
tiempos = [1.2, 0.8, 1.5, 0.9, 1.1, 2.3]
prom = sum(tiempos) / len(tiempos)
umbral = prom + 0.3
contador = 0
for t in tiempos:
    if t > umbral:
        contador += 1
print(prom, contador)

# 7
ventas = [120, 340, 210, 450, 380, 500]
categorias = []
for v in ventas:
    if v >= 400: categorias.append("alta")
    elif v >= 200: categorias.append("media")
    else: categorias.append("baja")

# 8
registros = [
    {"id":1,"valor":10,"activo":True},
    {"id":2,"valor":25,"activo":False},
    {"id":3,"valor":40,"activo":True},
    {"id":4,"valor":15,"activo":True},
]
suma = 0
for r in registros:
    if r["activo"]:
        suma += r["valor"]
print(suma)

# 9
m = [[1,2,3],[4,5,6]]
transpuesta = []
for j in range(len(m[0])):
    fila = []
    for i in range(len(m)):
        fila.append(m[i][j])
    transpuesta.append(fila)
print(transpuesta)

# 10
corpus = "data science data analysis data"
contador = 0
for palabra in corpus.split():
    if palabra == "data":
        contador += 1
print(contador)

# 11
a, b = 0, 1
while b <= 500:
    print(b)
    a, b = b, a + b

# 12
n = 987654321
suma = 0
while n > 0:
    suma += n % 10
    n //= 10
print(suma)

# 13
loss = 5.0
iteraciones = 0
while loss > 0.1:
    loss = loss * 0.9 - 0.05
    iteraciones += 1
print(iteraciones, loss)

# 14
valores = [10, 20, 30, -1, 50, 60]
i, suma, contados = 0, 0, 0
while i < len(valores) and valores[i] >= 0:
    suma += valores[i]
    contados += 1
    i += 1
print(suma, contados)

# 15
import random
objetivo = 100
acumulado = 0
pasos = 0
while acumulado < objetivo:
    acumulado += random.randint(1, 20)
    pasos += 1
print(pasos, acumulado)

# 16
def rango(lista):
    if not lista: return None
    mayor = menor = lista[0]
    for x in lista:
        if x > mayor: mayor = x
        if x < menor: menor = x
    return mayor - menor

rango_l = lambda l: None if not l else max(l) - min(l)

# 17
personas = [
    {"nombre":"Ana","edad":30,"salario":2500},
    {"nombre":"Luis","edad":25,"salario":3000},
    {"nombre":"Marta","edad":30,"salario":3200},
    {"nombre":"Pedro","edad":25,"salario":2800},
]
orden = sorted(personas, key=lambda p: (p["edad"], -p["salario"]))

# 18
filtrados = list(filter(lambda x: 0.4 <= x <= 0.9,
                        [0.15, 0.85, 0.42, 0.91, 0.33, 0.67]))

# 19
sumas = list(map(lambda t: t[0] + t[1], [(1,2),(3,4),(5,6)]))

# 20
def crear_pipeline(umbral, factor):
    return lambda v: "aceptado" if v * factor > umbral else "rechazado"

pipe = crear_pipeline(0.6, 1.5)
print(pipe(0.5))  # aceptado
print(pipe(0.3))  # rechazado
````
