## Sección I: Condicionales (Preguntas 1–5)

**3.** Escriba una función `es_correlacion_fuerte(r)` que retorne `True` si `|r| ≥ 0.7`, `False` en caso contrario. Use una **expresión ternaria**.

```python
# 1
def clasificar_edad(edad):
    if edad < 12: return "Niño"
    elif edad < 18: return "Adolescente"
    elif edad < 65: return "Adulto"
    else: return "Adulto mayor"

# 2
notas = [3.2, 4.5, 2.8, 4.9, 3.7, 1.5]
for n in notas:
    print("Aprobó" if n >= 3.0 else "Reprobó")

# 3
def es_correlacion_fuerte(r):
    return True if abs(r) >= 0.7 else False

# 4
modelos = {"lineal": 0.85, "árbol": 0.92, "knn": 0.78, "svm": 0.88}
mejor, mejor_val = None, -1
for nombre, acc in modelos.items():
    if acc > mejor_val:
        mejor, mejor_val = nombre, acc
print(f"Mejor modelo: {mejor} ({mejor_val})")

# 5
def es_nulo_o_vacio(valor):
    return valor is None or valor == "" or valor == [] or valor == {}

# 6
salarios = [2500, 1800, 3200, 4100, 1500]
prom = sum(salarios) / len(salarios)
contador = 0
for s in salarios:
    if s > prom:
        contador += 1
print(prom, contador)

# 7
puntuaciones = [88, 92, 79, 95, 85, 70]
normalizadas = []
for p in puntuaciones:
    normalizadas.append(p / 100)

# 8
clientes = [("Ana", 25), ("Luis", 40), ("Marta", 33), ("Pedro", 19)]
for nombre, edad in clientes:
    if edad > 30:
        print(nombre)

# 9
datos = [[1,2,3],[4,5,6],[7,8,9]]
plana = []
for fila in datos:
    for x in fila:
        plana.append(x)
print(plana)

# 10
texto = "machine learning con python"
contador = 0
for palabra in texto.split():
    if "a" in palabra:
        contador += 1
print(contador)

# 11
i = 2
while i <= 20:
    print(i)
    i += 2

# 12
n, fact = 8, 1
while n > 1:
    fact *= n
    n -= 1
print(fact)

# 13
n = 12345
digitos = 0
while n > 0:
    n //= 10
    digitos += 1
print(digitos)

# 14
error = 1.0
iteraciones = 0
while error > 0.01:
    error *= 0.8
    iteraciones += 1
print(iteraciones)

# 15
precios = [100, 200, 300, 0, 500, 600]
i, suma = 0, 0
while i < len(precios) and precios[i] != 0:
    suma += precios[i]
    i += 1
print(suma)

# 16
def promedio(lista):
    if not lista: return None
    return sum(lista) / len(lista)

prom = lambda lista: None if not lista else sum(lista) / len(lista)

# 17
empleados = [
    {"nombre": "Ana", "salario": 2500},
    {"nombre": "Luis", "salario": 3200},
    {"nombre": "Marta", "salario": 2100},
]
ordenados = sorted(empleados, key=lambda e: e["salario"], reverse=True)

# 18
impares = list(filter(lambda x: x % 2 != 0, [1, 5, 8, 12, 15, 20, 23]))

# 19
flotantes = list(map(lambda x: float(x), ["1.5", "2.7", "3.9", "4.1"]))

# 20
def crear_clasificador(umbral):
    return lambda valor: valor > umbral

clasif = crear_clasificador(0.5)
print(clasif(0.7))  # True
print(clasif(0.3))  # False
```
