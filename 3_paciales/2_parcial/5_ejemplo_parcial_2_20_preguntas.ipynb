# 11 — búsqueda binaria
def busqueda_binaria(lista, objetivo):
    if not lista: return -1
    izq, der = 0, len(lista) - 1
    while izq <= der:
        medio = (izq + der) // 2
        if lista[medio] == objetivo: return medio
        elif lista[medio] < objetivo: izq = medio + 1
        else: der = medio - 1
    return -1

# 16 — con def y lambda
def calcular_metricas(y_true, y_pred):
    if not y_true or len(y_true) != len(y_pred):
        return None
    tp = tn = fp = fn = 0
    for t, p in zip(y_true, y_pred):
        if t == 1 and p == 1: tp += 1
        elif t == 0 and p == 0: tn += 1
        elif t == 0 and p == 1: fp += 1
        else: fn += 1
    n = len(y_true)
    acc = (tp + tn) / n
    prec = tp / (tp + fp) if (tp + fp) > 0 else 0
    rec = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2*prec*rec / (prec + rec) if (prec + rec) > 0 else 0
    return {"accuracy": acc, "precision": prec, "recall": rec, "f1": f1}

# Versión lambda con dos lambdas anidadas
calcular_metricas_l = lambda y_t, y_p: None if not y_t else (lambda tp, fp, fn, n: {
    "accuracy": (tp + n - tp - fp - fn) / n,
    "precision": tp / (tp + fp) if (tp + fp) else 0,
    "recall": tp / (tp + fn) if (tp + fn) else 0,
    "f1": (lambda p, r: 2*p*r/(p+r) if (p+r) else 0)(
            tp / (tp + fp) if (tp + fp) else 0,
            tp / (tp + fn) if (tp + fn) else 0)
})(
    sum(1 for t, p in zip(y_t, y_p) if t == 1 and p == 1),
    sum(1 for t, p in zip(y_t, y_p) if t == 0 and p == 1),
    sum(1 for t, p in zip(y_t, y_p) if t == 1 and p == 0),
    len(y_t)
)

# 20 — pipeline
def crear_pipeline(*transformaciones):
    if not transformaciones:
        return lambda x: x
    return lambda x: reduce(lambda acc, f: f(acc), transformaciones, x)

# Alternativa sin reduce:
def crear_pipeline2(*tf):
    return lambda x: x if not tf else crear_pipeline2(*tf[1:])(tf[0](x))

cuadrar = lambda x: x ** 2
