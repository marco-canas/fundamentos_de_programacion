```python
# 1
def clasificar_balance(precision, recall):
    if not (0 <= precision <= 1 and 0 <= recall <= 1):
        return "entrada inválida"
    if precision < 0.5 or recall < 0.5:
        return "Crítico"
    if abs(precision - recall) > 0.2:
        return "Desbalanceado"
    if precision >= 0.8 and recall >= 0.8:
        return "Óptimo"
    return "Aceptable"

# 2
resultados = [(0.9,1),(0.4,0),(0.7,1),(0.6,0),
              (0.85,1),(0.3,0),(0.55,1),(0.95,1)]
for prob, real in resultados:
    pred = 1 if prob >= 0.5 else 0
    if pred == 1 and real == 1: print("TP")
    elif pred == 0 and real == 0: print("TN")
    elif pred == 1 and real == 0: print("FP")
    else: print("FN")

# 3
def detectar_tendencia(valores):
    if len(valores) < 2: return "sin datos"
    asc = desc = True
    for i in range(len(valores) - 1):
        if valores[i] >= valores[i+1]: asc = False
        if valores[i] <= valores[i+1]: desc = False
    if asc: return "ascendente"
    if desc: return "descendente"
    return "mixta"

# 4
reporte = {
    "modelo_A": {"accuracy": 0.85, "f1": 0.80, "activo": True},
    "modelo_B": {"accuracy": 0.90, "f1": 0.70, "activo": False},
    "modelo_C": {"accuracy": 0.78, "f1": 0.82, "activo": True},
}
mejor, mejor_f1 = None, -1
for nombre, m in reporte.items():
    if m["activo"] and m["f1"] > mejor_f1:
        mejor, mejor_f1 = nombre, m["f1"]
print(f"Mejor: {mejor}" if mejor else "Sin modelos activos")

# 5
def evaluar_umbral(valor, umbral, tolerancia=0.01):
    if valor is None or umbral is None:
        return "entrada inválida"
    if valor > umbral + tolerancia: return "sobre"
    if valor < umbral - tolerancia: return "bajo"
    return "en rango"

# 6 — una sola pasada
precios = [100,120,80,200,150,90,300]
n = len(precios)
suma = 0
for p in precios: suma += p
media = suma / n
var = 0
for p in precios: var += (p - media) ** 2
var /= n
sd = var ** 0.5
outliers = 0
for p in precios:
    if abs(p - media) > sd: outliers += 1
print(media, sd, outliers)
# Nota: la versión estricta "una sola pasada" usa Welford; aquí se
# acepta como equivalente si el estudiante lo justifica.

# 7
transacciones = [
    {"id":1,"monto":250,"moneda":"USD"},
    {"id":2,"monto":180,"moneda":"EUR"},
    {"id":3,"monto":420,"moneda":"USD"},
    {"id":4,"monto":90, "moneda":"COP"},
    {"id":5,"monto":310,"moneda":"USD"},
]
totales = {}
for t in transacciones:
    m = t["moneda"]
    totales[m] = totales.get(m, 0) + t["monto"]
print(totales)

# 8
m = [[1,2,3],[4,5,6]]
res = []
for j in range(len(m[0])):
    fila = []
    for i in range(len(m)):
        fila.append(m[i][j] * 2)
    res.append(fila)
print(res)

# 9 — una sola pasada
serie = [3,1,4,1,5,9,2,6,5,3,5]
max_v = min_v = serie[0]
max_i = min_i = 0
conteo = {}
for i, v in enumerate(serie):
    if v > max_v: max_v, max_i = v, i
    if v < min_v: min_v, min_i = v, i
    conteo[v] = conteo.get(v, 0) + 1
repetidos = sum(1 for v in conteo.values() if v >= 2)
print(max_v, max_i, min_v, min_i, repetidos)

# 10
texto = "machine learning es una rama de la inteligencia artificial"
frec = {}
for palabra in texto.split():
    frec[palabra] = frec.get(palabra, 0) + 1
top3 = sorted(frec.items(), key=lambda x: -x[1])[:3]
print(top3)

# 11
def es_primo(n):
    if n < 2: return False
    i = 2
    while i * i <= n:
        if n % i == 0: return False
        i += 1
    return True

candidato, iteraciones = 1001, 0
while True:
    iteraciones += 1
    if es_primo(candidato): break
    candidato += 1
print(candidato, iteraciones)

# 12
n = 123456789
invertido = 0
while n > 0:
    invertido = invertido * 10 + n % 10
    n //= 10
print(invertido)

# 13
centroide, nuevo = 0.0, 10.0
it = 0
while abs(nuevo - centroide) > 0.001:
    centroide = nuevo
    nuevo = (centroide + 5) / 2
    it += 1
print(it, centroide)

# 14 — mediana sin sorted
datos = [12,7,3,9,15,2,8,4]
copia = datos[:]
n = len(copia)
def extraer_min(lst):
    idx = 0
    for i in range(1, len(lst)):
        if lst[i] < lst[idx]: idx = i
    return lst.pop(idx)

mitad = n // 2
for _ in range(mitad):
    extraer_min(copia)
if n % 2 == 1:
    mediana = extraer_min(copia)
else:
    a = extraer_min(copia)
    b = extraer_min(copia)
    mediana = (a + b) / 2
print(mediana)

# 15
import random
N = 10000
dentro = 0
for _ in range(N):
    x, y = random.random(), random.random()
    if x*x + y*y <= 1:
        dentro += 1
pi_est = 4 * dentro / N
print(pi_est, abs(pi_est - 3.14159))

# 16
def varianza(lista):
    if len(lista) < 2: return None
    n = len(lista)
    media = sum(lista) / n
    return sum((x - media) ** 2 for x in lista) / n

varianza_l = lambda l: None if len(l) < 2 else \
    sum((x - sum(l)/len(l))**2 for x in l) / len(l)

# 17
modelos = [("A", 0.8, 0.9), ("B", 0.6, 0.6), ("C", 0.9, 0.7)]

def f1(t):
    _, p, r = t
    return 0 if (p + r) == 0 else 2 * p * r / (p + r)

orden = sorted(modelos, key=lambda t: f1(t), reverse=True)
# o directo:
orden = sorted(modelos,
    key=lambda t: 0 if (t[1]+t[2])==0 else 2*t[1]*t[2]/(t[1]+t[2]),
    reverse=True)

# 18
nums = [-2, 3, 4, -5, 6, 7, 8]
res = [(lambda x: x*x)(x) for x in nums if x > 0 and x % 2 == 0]
print(res)

# 19
from functools import reduce
v1 = [1,2,3,4]
v2 = [5,6,7,8]
producto = reduce(lambda a, b: a + b, map(lambda p: p[0]*p[1], zip(v1, v2)))
print(producto)

# 20
def crear_evaluador(umbrales):
    def etiqueta(v, umbrales=umbrales):
        for i, u in enumerate(umbrales):
            if v < u:
                return f"Q{i+1}"
        return f"Q{len(umbrales)+1}"
    return lambda v: etiqueta(v)

# Alternativa con lambda anidada
def crear_evaluador2(umbrales):
    return lambda v: next(
        (f"Q{i+1}" for i, u in enumerate(umbrales) if v < u),
        f"Q{len(umbrales)+1}"
    )

eval = crear_evaluador([0.25, 0.5, 0.75])
print(eval(0.1), eval(0.6), eval(0.9))
```
