16. Diseñe, desde cero, un programa en Python que resuelva la siguiente situación:  
    
Una veterinaria de Caucasia registra el peso (en kg) de las mascotas atendidas en el día. Algunos registros tienen errores de digitación: pesos negativos o valores None cuando la mascota se negó a subir a la báscula. Se necesita un programa que:    

*	Filtre los registros válidos (positivos y no None).
*	Calcule el peso promedio de las mascotas válidas.
*	Clasifique cada mascota válida como “pequeña” (< 5 kg), “mediana” (entre 5 y 20 kg) o “grande” (> 20 kg), usando operadores relacionales y lógicos.
*	Determine, usando un operador de pertenencia, si existe al menos una mascota con peso exactamente igual al promedio (comparación exacta de flotantes; puede discutir por qué esto es delicado).
*	Represente el resumen final como un diccionario con las claves "promedio", "pequeñas", "medianas", "grandes".  

Escriba el código completo, usando al menos: una lista de datos de prueba con errores incluidos, una lista por comprensión, un ciclo o comprensión con condicionales, y al menos tres tipos distintos de operadores estudiados en el curso (aritméticos, relacionales, lógicos, de pertenencia o de identidad). Comente brevemente cada bloque de su código.
Criterios de evaluación: corrección funcional (10 pts), uso apropiado y variado de estructuras/operadores del curso (6 pts), claridad y comentarios (4 pts).


Aquí tienes la solución completa redactada desde cero, incluyendo la discusión teórica sobre la comparación exacta de flotantes y el código comentado paso a paso.

---

### **Código en Python**

```python
# -------------------------------------------------------------------
# 1. DATOS DE PRUEBA
# -------------------------------------------------------------------
# Lista con registros de peso (kg) que incluye errores (None y números negativos)
registros_mascotas = [3.5, 12.0, None, -2.5, 25.0, 5.0, 12.0, None, 0.0, 22.5]

# -------------------------------------------------------------------
# 2. FILTRADO DE DATOS VÁLIDOS
# -------------------------------------------------------------------
# Usamos una lista por comprensión con condicional y operadores de identidad (is not)
# y relacionales (>) para obtener solo pesos positivos y distintos de None.
pesos_validos = [p for p in registros_mascotas if p is not None and p > 0]

# -------------------------------------------------------------------
# 3. CÁLCULO DEL PROMEDIO
# -------------------------------------------------------------------
# Uso de operadores aritméticos (+ implícito en sum, / para división)
if pesos_validos:
    promedio_peso = sum(pesos_validos) / len(pesos_validos)
else:
    promedio_peso = 0.0

# -------------------------------------------------------------------
# 4. CLASIFICACIÓN DE MASCOTAS
# -------------------------------------------------------------------
# Listas para almacenar la clasificación usando operadores relacionales (<, <=, >)
# y lógicos (and).
pequenas = [p for p in pesos_validos if p < 5]
medianas = [p for p in pesos_validos if p >= 5 and p <= 20]
grandes = [p for p in pesos_validos if p > 20]

# -------------------------------------------------------------------
# 5. OPERADOR DE PERTENENCIA Y DISCUSIÓN SOBRE FLOTANTES
# -------------------------------------------------------------------
# Uso del operador de pertenencia (in) para verificar si el promedio exacto está en la lista.
existe_coincidencia_exacta = promedio_peso in pesos_validos

# -------------------------------------------------------------------
# 6. RESUMEN FINAL EN DICCIONARIO
# -------------------------------------------------------------------
resumen_veterinaria = {
    "promedio": round(promedio_peso, 2),
    "pequeñas": pequenas,
    "medianas": medianas,
    "grandes": grandes,
}

# -------------------------------------------------------------------
# MOSTRAR RESULTADOS
# -------------------------------------------------------------------
print("=== RESUMEN VETERINARIA EN CAUCASIA ===")
print(f"Registros originales: {registros_mascotas}")
print(f"Registros válidos: {pesos_validos}")
print(f"Diccionario final: {resumen_veterinaria}")
print(
    f"¿Existe una mascota con peso EXACTO al promedio ({promedio_peso})?: {existe_coincidencia_exacta}"
)

```

---

### **Operadores utilizados en el script**

1. **Identidad:** `is not` (para validar que el dato no sea `None`).
2. **Relacionales:** `>`, `<`, `>=`, `<=` (para filtrar datos y clasificar categorías).
3. **Lógicos:** `and` (para definir el rango inclusivo de la categoría mediana).
4. **Aritméticos:** `/` (para calcular el promedio).
5. **Pertenencia:** `in` (para comprobar si el valor del promedio coincide exactamente con algún peso registrado).

### **Discusión teórica: ¿Por qué es delicada la comparación exacta de flotantes?**

La comparación directa con el operador de pertenencia `in` (que internamente evalúa la igualdad `==`) sobre números de coma flotante (`float`) es **delicada e impredecible** en programación por los siguientes motivos:

* **Representación binaria inexacta:** Los números flotantes se representan en hardware usando formato binario IEEE 754. Muchos números decimales simples (como `0.1` o `0.2`) no tienen una representación exacta en binario, lo que genera pequeños errores de redondeo al realizar operaciones aritméticas.
* **Sensibilidad de la división:** Al calcular `promedio_peso = sum(...) / len(...)`, el resultado suele tener múltiples decimales imprecisos por la división (por ejemplo, `13.583333333333334`). A menos que una mascota haya registrado exactamente esa misma secuencia decimal impredecible, la comparación `promedio_peso in pesos_validos` devolverá `False`, incluso si en teoría matemática el promedio coincidiera.
* **Buenas prácticas:** En aplicaciones reales, en lugar de una igualdad estricta (`==` o `in`), se suele comprobar si la diferencia absoluta entre el flotante y el valor objetivo es menor a un umbral de tolerancia $\epsilon$ (por ejemplo, usando `math.isclose()` o `abs(a - b) < 1e-9`).
