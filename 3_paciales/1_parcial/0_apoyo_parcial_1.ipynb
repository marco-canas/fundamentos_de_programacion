## **Código en Python**


````python
# -------------------------------------------------------------------
# 1. DATOS DE PRUEBA
# -------------------------------------------------------------------
# Lista con registros de peso (kg) que incluye errores (None y números negativos)
registros_mascotas = [3.5, 12.0, None, -2.5, 25.0, 5.0, 12.0, None, 0.0, 22.5]

# -------------------------------------------------------------------
# 2. FILTRADO DE DATOS VÁLIDOS
# -------------------------------------------------------------------
# Usamos una lista por comprensión con condicional y operadores de identidad (is not)
# y relacionales (>) para obtener solo pesos positivos y distintos de None.
pesos_validos = [p for p in registros_mascotas if p is not None and p > 0]

# -------------------------------------------------------------------
# 3. CÁLCULO DEL PROMEDIO
# -------------------------------------------------------------------
# Uso de operadores aritméticos (+ implícito en sum, / para división)
promedio_peso = sum(pesos_validos) / len(pesos_validos)

# -------------------------------------------------------------------
# 4. CLASIFICACIÓN DE MASCOTAS
# -------------------------------------------------------------------
# Listas para almacenar la clasificación usando operadores relacionales (<, <=, >)
# y lógicos (and).
pequenas = [p for p in pesos_validos if p < 5]
medianas = [p for p in pesos_validos if p >= 5 and p <= 20]
grandes = [p for p in pesos_validos if p > 20]

# -------------------------------------------------------------------
# 5. OPERADOR DE PERTENENCIA Y DISCUSIÓN SOBRE FLOTANTES
# -------------------------------------------------------------------
# Uso del operador de pertenencia (in) para verificar si el promedio exacto está en la lista.
existe_coincidencia_exacta = promedio_peso in pesos_validos

# -------------------------------------------------------------------
# 6. RESUMEN FINAL EN DICCIONARIO
# -------------------------------------------------------------------
resumen_veterinaria = {
    "promedio": round(promedio_peso, 2),
    "pequeñas": pequenas,
    "medianas": medianas,
    "grandes": grandes,
}

# -------------------------------------------------------------------
# MOSTRAR RESULTADOS
# -------------------------------------------------------------------
print("=== RESUMEN VETERINARIA EN CAUCASIA ===")
print(f"Registros originales: {registros_mascotas}")
print(f"Registros válidos: {pesos_validos}")
print(f"Diccionario final: {resumen_veterinaria}")
print(
    f"¿Existe una mascota con peso EXACTO al promedio ({promedio_peso})?: {existe_coincidencia_exacta}"
)
````


### **Operadores utilizados en el script**

1. **Identidad:** `is not` (para validar que el dato no sea `None`).
2. **Relacionales:** `>`, `<`, `>=`, `<=` (para filtrar datos y clasificar categorías).
3. **Lógicos:** `and` (para definir el rango inclusivo de la categoría mediana).

# **Discusión teórica: ¿Por qué es delicada la comparación exacta de flotantes?**



La comparación directa con el operador de pertenencia `in` (que internamente evalúa la igualdad `==`) sobre números de coma flotante (`float`) es **delicada e impredecible** en programación por los siguientes motivos:



* **Representación binaria inexacta:** Los números flotantes se representan en hardware usando formato binario IEEE 754. Muchos números decimales simples (como `0.1` o `0.2`) no tienen una representación exacta en binario, lo que genera pequeños errores de redondeo al realizar operaciones aritméticas.


* **Sensibilidad de la división:** Al calcular `promedio_peso = sum(...) / len(...)`, el resultado suele tener múltiples decimales imprecisos por la división (por ejemplo, `13.583333333333334`). A menos que una mascota haya registrado exactamente esa misma secuencia decimal impredecible, la comparación `promedio_peso in pesos_validos` devolverá `False`, incluso si en teoría matemática el promedio coincidiera.


* **Buenas prácticas:** En aplicaciones reales, en lugar de una igualdad estricta (`==` o `in`), se suele comprobar si la diferencia absoluta entre el flotante y el valor objetivo es menor a un umbral de tolerancia $\epsilon$ (por ejemplo, usando `math.isclose()` o `abs(a - b) < 1e-9`).

````python
promedio_esta_en_pesos_mascotas = any([abs(peso_promedio - p) < 0.001 for p in lista_filtrada])
promedio_esta_en_pesos_mascotas
````

b) **Explicación del tipo `float`:**
Al calcular una raíz cuadrada mediante la elevación a una potencia decimal (`** 0.5`), Python evalúa la operación dentro del dominio de los números reales (flotantes). Por diseño del lenguaje, la operación de potenciación con exponentes no enteros o fraccionarios retorna siempre un tipo de dato `float` para poder representar aproximaciones continuas, sin importar si el valor numérico final coincide con un entero exacto ($17.0$).


El término **aproximación continua** responde a cómo las computadoras representan la matemática del mundo real mediante el estándar de punto flotante (IEEE 754).

A diferencia de los números enteros (`int`), que son **discretos** y saltan de un valor exacto a otro (como 1, 2, 3), los números reales pertenecen a un **continuo matemático**: entre cualquier par de números reales existen infinitos números intermedios.

Cuando Python se enfrenta a una potencia fraccionaria como `** 0.5`, ocurren dos razones clave para que el sistema opere bajo esta lógica:

* **Soporte para irracionales y decimales:** La inmensa mayoría de las raíces cuadradas no resultan en números enteros exactos (por ejemplo, $\sqrt{2} \approx 1.41421356...$). Como los enteros de Python no pueden guardar decimales, la operación debe ejecutarse en un tipo de dato que permita "moverse" de forma continua a lo largo de la recta numérica.
* **Aproximación por límites de hardware:** Un ordenador no puede almacenar infinitos decimales para representar un número irracional. Por ello, el tipo `float` aproxima ese continuo matemático usando 64 bits de memoria. Aunque la raíz de $289$ sea exactamente el entero $17$, el algoritmo de potencia decimal no sabe de antemano si el resultado será exacto o un decimal infinito.

Para mantener la **consistencia de tipos** en el lenguaje, Python evalúa toda la operación en el dominio flotante. La máquina calcula una **aproximación en un espacio continuo** y entrega `17.0` (un flotante con precisión decimal), evitando así que el tipo de retorno de una función cambie dinámicamente según los números ingresados.

# B2. El siguiente código pretende verificar el acceso de un usuario a una plataforma, y por qué es un ejemplo típico de mal uso del argumento por defecto mutable:

````python
def registrar_intento(usuario, historial=[]):
    historial.append(usuario)
    return historial

print(registrar_intento("ana"))
print(registrar_intento("luis"))
````

**Resultado de la segunda llamada**

La segunda llamada imprime:
`['ana', 'luis']`

*(El autor probablemente esperaba que imprimiera solo `['luis']`)*.

---



**Explicación (Mutabilidad e Identidad de Objetos)**

En Python, las definiciones de funciones son código ejecutable y **los valores por defecto de los argumentos se evalúan una sola vez**, en el momento en que la función es definida (al cargar el script), no cada vez que se invoca.

* La lista vacía `[]` se crea un objeto único en memoria y se asigna al parámetro `historial`.
* Como las listas son objetos **mutables**, la llamada `.append("ana")` modifica esa misma instancia almacenada en memoria.
* Cuando se llama a `registrar_intento("luis")` sin pasar un segundo argumento, Python reutiliza exactamente **el mismo objeto lista** (`id(historial)` no cambia), acumulando las llamadas previas en lugar de iniciar con una lista limpia.

---

**Corrección estándar**

La forma idónea en Python para evitar la acumulación no deseada es utilizar `None` como valor por defecto y crear una nueva lista dentro de la función si no se proporciona una:



````python
def registrar_intento(usuario, historial=None):
    if historial is None:
        historial = []
    historial.append(usuario)
    return historial

print(registrar_intento("ana"))   # Imprime: ['ana']
print(registrar_intento("luis"))  # Imprime: ['luis']
````
