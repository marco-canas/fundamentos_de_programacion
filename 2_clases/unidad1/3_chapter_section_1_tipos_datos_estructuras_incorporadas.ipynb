## 3.1 Tipos de Datos en Python

### Desarrollo teórico

Python es un lenguaje de **tipado dinámico**: no se declara el tipo de una variable explícitamente, el intérprete lo infiere en tiempo de ejecución. Para ciencia de datos esto es clave porque **pandas y NumPy heredan y extienden estos tipos base** (por ejemplo, `int64`, `float64`, `object`, `bool` en un DataFrame provienen directamente de los tipos nativos de Python y C).

| Tipo Python | Ejemplo | Uso típico en ciencia de datos |
|---|---|---|
| `int` | `edad = 28` | Conteos, IDs, categorías codificadas |
| `float` | `precio = 19.99` | Variables numéricas continuas, métricas |
| `str` | `nombre = "Ana"` | Texto, categorías, NLP |
| `bool` | `activo = True` | Flags, filtros booleanos, máscaras |
| `complex` | `z = 3 + 4j` | Procesamiento de señales, FFT |
| `NoneType` | `valor = None` | Representar datos faltantes (antes de `NaN`) |



````python
edad = 28              # int
precio = 19.99         # float
ciudad = "Medellín"    # str
activo = True          # bool
sin_dato = None        # NoneType

print(type(edad), type(precio), type(ciudad), type(activo), type(sin_dato))
````

**Dato clave para ciencia de datos:** cuando pandas lee un CSV con celdas vacías, esas celdas se convierten en `NaN` (de NumPy), que se comporta de forma distinta a `None`. Vale la pena mencionarlo pronto para evitar confusiones futuras.


````python
import math
print(None == None)          # True
print(math.nan == math.nan)  # False -> NaN nunca es igual a sí mismo
````

## 3.2 Variables en Python

### Desarrollo teórico

Una variable en Python es una **etiqueta (referencia)** que apunta a un objeto en memoria, no una "caja" que contiene el valor directamente. Esto es fundamental para entender más adelante la diferencia entre mutabilidad e inmutabilidad.

## 3.3 Estructuras de Datos Incorporadas en Python (Mutables e Inmutables)

### Desarrollo teórico

En Python, cada estructura de datos se clasifica según si su contenido puede modificarse **después de creado**:

| Categoría | Estructuras | ¿Se puede modificar? |
|---|---|---|
| **Mutables** | `list`, `dict`, `set` | Sí, sin cambiar su identidad (`id()`) |
| **Inmutables** | `tuple`, `str`, `int`, `float`, `frozenset` | No, cualquier "cambio" crea un nuevo objeto |



````python
lista = [1, 2, 3]
print(id(lista))
lista.append(4)
print(id(lista))   # mismo id -> es mutable

tupla = (1, 2, 3)
print(id(tupla))
tupla = tupla + (4,)
print(id(tupla))   # id distinto -> se creó un objeto nuevo
````

**Por qué importa en ciencia de datos:**
- Las estructuras **inmutables** (tuplas) son ideales como *llaves de diccionario* o para representar registros fijos (ej. coordenadas geográficas, filas inmutables).
- Las estructuras **mutables** (listas, sets) son la base para acumular resultados, limpiar datos y construir *pipelines* de transformación.
- Un error muy común en notebooks es pasar una lista como valor por defecto en una función y modificarla sin darse cuenta, afectando a todas las llamadas siguientes.



````python
def agregar_registro(registro, historial=[]):  # mala práctica
    historial.append(registro)
    return historial

print(agregar_registro("A"))  # ['A']
print(agregar_registro("B"))  # ['A', 'B']  <- efecto inesperado por mutabilidad
````

## 3.3.1 Listas

### Desarrollo

Las listas son colecciones **ordenadas, mutables y que permiten duplicados**. Son la estructura más usada al leer datos crudos antes de convertirlos en estructuras más especializadas (arrays de NumPy, Series de pandas).



````python
temperaturas = [22.5, 23.1, 21.8, 24.0, 22.9]

temperaturas.append(25.3)          # agregar al final
temperaturas.insert(0, 20.0)       # insertar en posición
temperaturas.remove(21.8)          # eliminar por valor
promedio = sum(temperaturas) / len(temperaturas)

print(temperaturas)
print(f"Promedio: {promedio:.2f}")

# Slicing, muy usado en preprocesamiento
print(temperaturas[:3])   # primeros 3 registros
print(temperaturas[-2:])  # últimos 2 registros

# List comprehension (esencial en ciencia de datos)
en_fahrenheit = [t * 9/5 + 32 for t in temperaturas]
print(en_fahrenheit)
````

# Ejemplo aplicado a ciencia de datos


````python
def evaluar_modelo(y_real, y_pred):
    aciertos = sum(1 for r, p in zip(y_real, y_pred) if r == p)
    accuracy = aciertos / len(y_real)
    return accuracy, aciertos, len(y_real)   # función que retorna una tupla

y_real = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 0]

acc, correctos, total = evaluar_modelo(y_real, y_pred)
print(f"Accuracy: {acc:.2f} ({correctos}/{total})")
````

## 3.3.3 Conjuntos (Sets)

### Desarrollo

Los conjuntos son colecciones **desordenadas, mutables y sin elementos duplicados**. En ciencia de datos son extremadamente útiles para **deduplicar datos** y realizar operaciones de conjuntos (intersección, unión, diferencia) entre categorías o grupos de usuarios.



````python
categorias_tienda_a = {"electrónica", "ropa", "hogar", "juguetes"}
categorias_tienda_b = {"ropa", "hogar", "libros"}

print(categorias_tienda_a & categorias_tienda_b)  # intersección
print(categorias_tienda_a | categorias_tienda_b)  # unión
print(categorias_tienda_a - categorias_tienda_b)  # diferencia
````

# Ejemplo aplicado a ciencia de datos



````python
# Deduplicar correos de una campaña de marketing
correos = ["a@mail.com", "b@mail.com", "a@mail.com", "c@mail.com", "b@mail.com"]

correos_unicos = set(correos)
print(f"De {len(correos)} registros, hay {len(correos_unicos)} correos únicos")

# Clientes que compraron en ambos meses (intersección de conjuntos)
clientes_enero = {"C1", "C2", "C3", "C4"}
clientes_febrero = {"C2", "C4", "C5"}
clientes_fieles = clientes_enero & clientes_febrero
print("Clientes recurrentes:", clientes_fieles)
````

## 3.3.4 Diccionarios

### Desarrollo

Los diccionarios almacenan pares **clave-valor**, son mutables y (desde Python 3.7) mantienen el orden de inserción. Son la estructura más cercana a un **registro JSON** o a una fila con nombres de columna, por lo que son la base conceptual de un `DataFrame` de pandas (que internamente puede pensarse como un diccionario de columnas).



````python
cliente = {
    "id": 101,
    "nombre": "Camila",
    "edad": 34,
    "compras": [230000, 150000, 89000]
}

print(cliente["nombre"])
cliente["ciudad"] = "Bogotá"       # agregar clave
cliente["edad"] = 35               # actualizar valor
del cliente["compras"]             # eliminar clave

for clave, valor in cliente.items():
    print(f"{clave}: {valor}")
````

# Ejemplo aplicado a ciencia de datos



````python
# Representar un mini-dataset como lista de diccionarios (formato típico de JSON/API)
dataset = [
    {"nombre": "Ana",   "edad": 28, "ciudad": "Medellín"},
    {"nombre": "Luis",  "edad": 35, "ciudad": "Bogotá"},
    {"nombre": "Marta", "edad": 41, "ciudad": "Cali"},
]

edades = [registro["edad"] for registro in dataset]
print("Edad promedio:", sum(edades) / len(edades))

# Conteo de frecuencia por ciudad (patrón muy común en EDA)
conteo_ciudades = {}
for registro in dataset:
    ciudad = registro["ciudad"]
    conteo_ciudades[ciudad] = conteo_ciudades.get(ciudad, 0) + 1

print(conteo_ciudades)
````
