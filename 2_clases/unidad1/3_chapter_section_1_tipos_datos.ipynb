### Sesión 2 — Conjuntos y diccionarios



**Pregunta orientadora:**
¿Puede Python representar estructuras matemáticas como conjuntos y funciones?



````python
A = {1,2,3,4}
B = {3,4,5,6}

A.union(B), A.intersection(B), A.difference(B)
````

````python
# Diccionario como función
f = {1:1, 2:4, 3:9, 4:16} # un conjunto de parejas clave:valor
````

````python
# Cómo accede a la imagen de una preimagen 
f[4]
````

**Interpretación**

* Un diccionario como función discreta.
* Dominio y codominio computacional.

### Sesión 3 — Aplicación a ciencia de datos



**Problema:**
Registrar temperaturas diarias y analizarlas.


````python
temperaturas = [30,31,29,28,32,33,30]
dias = ["L","M","Mi","J","V","S","D"]
parejas_dias_temperatura = list(zip(dias, temperaturas))
registro = dict(parejas_dias_temperatura)
list(parejas_dias_temperatura), registro
````

````python
# Análisis básico
max(temperaturas), min(temperaturas), sum(temperaturas)/len(temperaturas)
````
