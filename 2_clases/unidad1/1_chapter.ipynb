````python
import qrcode
from IPython.display import display
from PIL import Image

def generar_qr(url, version=1, box_size=4, border=4, fill_color="black", back_color="white"):
    """
    Genera y muestra un código QR para una URL dada en un Jupyter Notebook.
    
    Parámetros:
    - url: La URL para la cual generar el código QR
    - version: Tamaño del QR (1-40, donde 1 es el más pequeño)
    - box_size: Número de píxeles por cada "caja" del QR
    - border: Grosor del borde blanco alrededor del QR (en cajas)
    - fill_color: Color del código QR
    - back_color: Color de fondo del código QR
    
    Retorna:
    - Muestra el código QR directamente en el notebook
    - Retorna el objeto QR generado
    """
    # Configurar el generador de QR
    qr = qrcode.QRCode(
        version=version,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=box_size,
        border=border,
    )
    
    # Añadir la URL al QR
    qr.add_data(url)
    qr.make(fit=True)
    
    # Crear la imagen del QR
    img = qr.make_image(fill_color=fill_color, back_color=back_color)
    
    return img

````

## 1️. Programación **procedimental**

- El código se ejecuta **paso a paso**, usando funciones simples y variables globales.



**Algoritmo procedimental para graficar la función $y = x^2$**

1. **Importar las bibliotecas necesarias**  
   - Se importa `matplotlib.pyplot` para generar gráficas.  
   - Se importa `numpy` para realizar operaciones matemáticas y crear arreglos numéricos.



2. **Generar los datos**  
   - Se crea un arreglo llamado `x` que contiene 100 números igualmente espaciados, comenzando en -10 y terminando en 10. (Progresión aritmética de primer y último térmono dados y además el número de términos de la progresión dados) 
   - Se calcula el cuadrado de cada valor de `x` y se almacena en un arreglo llamado `y`.



3. **Configurar la gráfica**  
   - Se crea una nueva figura con un tamaño de 6 unidades de ancho por 4 de alto.  
   - Se asigna un título a la gráfica que indica que se trata del paradigma procedimental y muestra la fórmula `y = x²`.  
   - Se etiquetan los ejes horizontal y vertical con "x" y "y", respectivamente.  
   - Se establecen las marcas del eje x de -10 a 10, de uno en uno.  
   - Se activa una cuadrícula de fondo para facilitar la lectura.



4. **Dibujar la gráfica**  
   - Se traza la curva uniendo los puntos generados por `x` y `y` con líneas continuas.  
   - Se añade una leyenda que muestra la expresión $y = x^{2}$.



5. **Guardar y mostrar el resultado**  
   - Se guarda la gráfica como una imagen en la carpeta "Descargas" del usuario, con el nombre `paradigma_procedimental.png`.  
   - Finalmente, se muestra la gráfica en pantalla.

````python
import matplotlib.pyplot as plt
import numpy as np

# Datos
x = np.linspace(start = -10, stop =  10, num = 100) # crea una progresión aritmética dados el primero y último términos y el número de términos 
y = x**2   # Cálcula los cuadrados de cada término de la prograsión aritmética. 

# Gráfica
plt.figure(figsize = (6,4))
plt.title(r"Paradigma Procedimental: $y = x^2$")
plt.xlabel("x")
plt.ylabel("y")
plt.xticks(np.arange(-10, 11))
plt.grid(alpha = 0.5, linestyle = "--")
plt.plot(x, y)   # Traza los puntos y une con líneas o segmentos 
plt.legend([r"$y = x^2$"])  # Agrega una leyenda a la gráfica
plt.savefig(r"C:\Users\marco\Downloads\paradigma_procedimental.png")
plt.show()
````

* El flujo del programa es secuencial
* Es el paradigma más cercano al “algoritmo en papel”

- Ideal para **introducción a la programación** y pensamiento algorítmico.




# 2. Programación **funcional**

- Usamos **funciones** para tener código reutilizable. 



````python
import matplotlib.pyplot as plt
import numpy as np

def funcion_cuadratica(x):
    return x**2

def graficar(func, x):
    """Función para graficar una función matemática.
    Parámetros: 
    - func: La función matemática a graficar 
    - x: Los valores del dominio para los cuales se evaluará la función, 
    cómo una progresión aritmética. 
    """
    plt.figure() # Genera el lienzo o cuadra para hacer el gráfico 
    plt.title(r"Paradigma Funcional: $y = x^2$")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(alpha = 0.3, linestyle = "--")
    plt.plot(x, func(x))
    plt.legend([r"$y = x^2$"])  # Agrega una leyenda a la gráfica
    plt.savefig(r"C:\Users\marco\Downloads\paradigma_funcional.png")
    plt.show()

x = np.linspace(-10, 10, 100) # Genera elgunos puntos del dominio 
graficar(funcion_cuadratica, x)
````

- **Idea clave:**

* Las funciones son **ciudadanos de primera clase**
* Una función recibe otra función como argumento
* No hay modificación de estados

- Ideal para **pensamiento matemático**, **transformaciones** y **ciencia de datos**.

#   - Interactive 

El modo interactivo de Python Shell ofrece la función Read, Eval Print, and Loop (REPL), que proporciona retroalimentación interactiva instantánea al usuario. 


## ¿Qué es REPL?

Cada vez que ves el prompt:

```text
>>>
```



Python está haciendo continuamente:

1. **Read** → lee lo que escribes
2. **Eval** → lo evalúa
3. **Print** → muestra el resultado
4. **Loop** → vuelve a esperar otra instrucción

---



## 1️. Ejemplo básico: evaluación inmediata

### En el Python Shell


````python

```python
>>> 2 + 3
5
```
````


- **REPL**

* Read: `2 + 3`
* Eval: calcula la suma
* Print: `5`
* Loop: vuelve a `>>>`

- No necesitas `print()` → el intérprete imprime automáticamente el resultado.
