## Problema común

Graficar la función
$$
f(x) = x^2
$$
en el intervalo $[-4, 4]$

## 1️. Programación **procedimental**

- El código se ejecuta **paso a paso**, usando funciones simples y variables globales.



**Algoritmo procedimental para graficar la función $y = x^2$**

1. **Importar las bibliotecas necesarias**  
   - Se importa `matplotlib.pyplot` para generar gráficas.  
   - Se importa `numpy` para realizar operaciones matemáticas y crear arreglos numéricos.



2. **Generar los datos**  
   - Se crea un arreglo llamado `x` que contiene 100 números igualmente espaciados, comenzando en -10 y terminando en 10. (Progresión aritmética de primer y último térmono dados y además el número de términos de la progresión dados) 
   - Se calcula el cuadrado de cada valor de `x` y se almacena en un arreglo llamado `y`.



3. **Configurar la gráfica**  
   - Se crea una nueva figura con un tamaño de 6 unidades de ancho por 4 de alto.  
   - Se asigna un título a la gráfica que indica que se trata del paradigma procedimental y muestra la fórmula `y = x²`.  
   - Se etiquetan los ejes horizontal y vertical con "x" y "y", respectivamente.  
   - Se establecen las marcas del eje x de -10 a 10, de uno en uno.  
   - Se activa una cuadrícula de fondo para facilitar la lectura.



4. **Dibujar la gráfica**  
   - Se traza la curva uniendo los puntos generados por `x` y `y` con líneas continuas.  
   - Se añade una leyenda que muestra la expresión $y = x^{2}$.



5. **Guardar y mostrar el resultado**  
   - Se guarda la gráfica como una imagen en la carpeta "Descargas" del usuario, con el nombre `paradigma_procedimental.png`.  
   - Finalmente, se muestra la gráfica en pantalla.

````python
import matplotlib.pyplot as plt
import numpy as np

# Datos
x = np.arange(start = -4, stop =  4 + 1, step = 1) # crea una progresión aritmética dados el primero y último términos y el número de términos 
y = x**2   # Cálcula los cuadrados de cada término de la progresión aritmética. 

# Gráfica
plt.figure(figsize = (6,4))
plt.title(r"Paradigma Procedimental: $y = x^2$")
plt.xlabel("x")
plt.ylabel("y")
plt.xticks(np.arange(-4, 5))
plt.grid(alpha = 0.5, linestyle = "--")
plt.plot(x, y)   # Traza los puntos y une con líneas o segmentos 
plt.legend([r"$y = x^2$"])  # Agrega una leyenda a la gráfica
plt.savefig(r"C:\Users\marco\Downloads\paradigma_procedimental.png")
plt.show()
````

# 2. Programación **funcional**

- Usamos **funciones** para tener código reutilizable. 



````python
import matplotlib.pyplot as plt
import numpy as np

def funcion_cuadratica(x):
    return x**2

def graficar(func, x):
    """Función para graficar una función matemática.
    Parámetros: 
    - func: La función matemática a graficar 
    - x: Los valores del dominio para los cuales se evaluará la función, 
    cómo una progresión aritmética. 
    """
    plt.figure() # Genera el lienzo o cuadra para hacer el gráfico 
    plt.title(r"Paradigma Funcional: $y = x^2$")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(alpha = 0.3, linestyle = "--")
    plt.plot(x, func(x))
    plt.legend([r"$y = x^2$"])  # Agrega una leyenda a la gráfica
    plt.savefig(r"C:\Users\marco\Downloads\paradigma_funcional.png")
    plt.show()

x = np.arange(start = -4, stop = 4 + 1, step = 1) # Genera algunos puntos del dominio de graficación
graficar(funcion_cuadratica, x) # llama a la función que traza el gráfico de la función cuadrática
````

- **Idea clave:**

* Las funciones son **ciudadanos de primera clase**
* Una función recibe otra función como argumento
* No hay modificación de estados

- Ideal para **pensamiento matemático**, **transformaciones** y **ciencia de datos**.

# 3. Programación **orientada a objetos (POO)**

- Encapsulamos **datos + funciones que actuan sobre los datos** en una clase.



````python
import matplotlib.pyplot as plt # Hacer visualizaciones 
import numpy as np              # Es una librería o colección de funciones matemáticas 

class GraficaParabola:
    def __init__(self, xmin=-4, xmax=4, step = 1):
        self.x = np.arange(xmin, xmax + 1, step)
        self.y = self.x**2
        self.xmin = xmin
        self.xmax = xmax
    def graficar(self):
        plt.figure(figsize = (6,4))
        plt.title(r"Paradigma Orientado a Objetos: $y = x^2$")
        plt.vlines(x = 0, ymin = np.min(self.y), ymax=np.max(self.y), 
                   colors='gray', linestyles='dashed')  # Línea vertical en x=0
        plt.hlines(y = 0, xmin=self.xmin, xmax=self.xmax, colors='gray', linestyles='dashed')  # Línea horizontal en y=0
        plt.xlabel("x")
        plt.ylabel("y")
        plt.grid(True)
        plt.xticks(np.arange(self.xmin, self.xmax + 1, 2))
        plt.plot(self.x, self.y)
        plt.savefig(r"C:\Users\marco\Downloads\paradigma_de_orientación_a_objetos.png")
        plt.show()

# Uso del objeto
parabola = GraficaParabola()
parabola.graficar()
````

- **Idea clave:**

* La gráfica es un **objeto**
* Los datos (`x`, `y`) y el método (`graficar`) viven juntos
* Facilita reutilización, extensión y mantenimiento

- Ideal para **modelado**, **simulación** y **proyectos grandes**.

# - Interpreted 

Python se considera un lenguaje interpretado porque, a diferencia de lenguajes compilados como C o C++, no se traduce directamente a código máquina antes de ejecutarse. En su lugar, el código se procesa y ejecuta línea por línea (o de forma más precisa, instrucción por instrucción) por un programa llamado **intérprete**, en el momento de correr el programa.

## Cómo funciona en la práctica

En realidad, el proceso es un poco más matizado que "leer línea por línea":

1. **Compilación a bytecode**: Cuando ejecutas un script de Python, el intérprete primero compila el código fuente a un formato intermedio llamado *bytecode* (los archivos `.pyc` que a veces ves en la carpeta `__pycache__`).
2. **Ejecución por la máquina virtual**: Ese bytecode no es código máquina nativo, sino instrucciones que entiende la **PVM (Python Virtual Machine)**, que es la que realmente interpreta y ejecuta cada instrucción sobre la marcha.

Este paso intermedio ocurre automáticamente y de forma transparente para el programador, por eso a Python se le sigue llamando "interpretado" en vez de "compilado", aunque técnicamente tenga una etapa de compilación interna.

## Por qué esto importa (diferencias prácticas)

- **No necesitas compilar manualmente**: escribes tu código y lo ejecutas directamente con `python archivo.py`, sin un paso previo de compilación explícito como en C (`gcc archivo.c -o programa`).
- **Portabilidad**: el mismo bytecode puede ejecutarse en cualquier sistema que tenga instalado el intérprete de Python correspondiente, sin necesidad de recompilar para cada arquitectura.
- **Errores en tiempo de ejecución**: muchos errores (como los de tipo) solo se detectan cuando el intérprete llega a esa línea durante la ejecución, no antes.
- **Velocidad**: generalmente es más lento que un lenguaje compilado a código máquina nativo, porque hay una capa adicional (la máquina virtual) traduciendo instrucciones mientras el programa corre.

## Una aclaración importante

Decir que "Python es un lenguaje interpretado" es en realidad una simplificación. Más preciso sería decir que **la implementación estándar de Python (CPython) usa un modelo híbrido** de compilación a bytecode + interpretación. De hecho, existen otras implementaciones como **PyPy**, que usa compilación *just-in-time* (JIT) para acelerar la ejecución, acercándose más al rendimiento de lenguajes compilados.

En resumen: se le llama interpretado porque el usuario no compila explícitamente a código máquina, y la ejecución depende de un intérprete que traduce y corre las instrucciones en tiempo real, aunque internamente use bytecode como paso intermedio.

#   - Interactive 

El modo interactivo de Python: Windows PowerShell ofrece la función Read, Eval, Print, and Loop (REPL), que proporciona retroalimentación interactiva instantánea al usuario. 


## ¿Qué es REPL?

Cada vez que ves el prompt:

```text
>>>
```



Python está haciendo continuamente:

1. **Read** → lee lo que escribes
2. **Eval** → lo evalúa
3. **Print** → muestra el resultado
4. **Loop** → vuelve a esperar otra instrucción

---



## 1️. Ejemplo básico: evaluación inmediata de Python a través de la terminal



### En el Python Shell


````python

```python
>>> 2 + 3
5
```
````


- **REPL**

* Read: `2 + 3`
* Eval: calcula la suma
* Print: `5`
* Loop: vuelve a `>>>`

- No necesitas `print()` → el intérprete imprime automáticamente el resultado.

# Ejercicios según la metodología de Aprendizaje Basado en Problemas (ABP)


**1. El detective del lenguaje**
*Situación:* Un colega afirma que "Python es simplemente C pero más lento". Tu tarea es escribir un informe corto (con código de evidencia) que confirme o refute esa afirmación, apoyándote en los conceptos de *interpretado*, *alto nivel* y *bytecode* vistos en el capítulo.
*Evalúa:* comprensión conceptual de qué significa "interpretado", capacidad de argumentar con evidencia técnica.



**2. Migración de código legado**
*Situación:* Una empresa ficticia tiene scripts en Python 2 (`print "hola"`) y quiere saber si vale la pena migrarlos a Python 3.14. Investiga las diferencias clave entre Python 2 y 3 y redacta una recomendación justificada.
*Evalúa:* investigación autónoma sobre versiones, pensamiento crítico aplicado a decisiones reales.



**3. Elegir la herramienta correcta**
*Situación:* Tres equipos (uno de bases de datos, uno de ciencia de datos, uno de desarrollo web) deben elegir un lenguaje para un proyecto nuevo. Argumenta cuándo Python **es** la mejor opción y cuándo **no lo es**, usando las características de propósito general, multiparadigma y tipado dinámico.
*Evalúa:* transferencia del conocimiento a contextos de decisión profesional.



**4. Un mismo problema, tres formas de resolverlo**
*Situación:* Recibes el código procedimental que grafica $y=x^2$ (celda 21). Refactóralo primero a estilo **funcional** y luego a **orientado a objetos**, siguiendo los ejemplos del cuaderno, y explica en Markdown qué ventaja aporta cada refactorización.
*Evalúa:* comprensión profunda de los tres paradigmas, no memorística sino aplicada.



**5. El intérprete como laboratorio**
*Situación:* Un estudiante nuevo no entiende por qué en Jupyter no necesita escribir `print()` para ver un resultado. Diseña una mini-demostración en el REPL (como las de las celdas 64–89) que le enseñe el ciclo *Read-Eval-Print-Loop* con al menos un ejemplo de error controlado (`ZeroDivisionError`).
*Evalúa:* dominio del entorno interactivo y capacidad de enseñar (metacognición).



**6. Sistema de registro con validación de tipos**
*Situación:* Debes construir un mini "formulario" en Python que capture nombre (cadena), edad (entero), promedio (flotante) y si el estudiante está activo (booleano), e imprima el `type()` de cada uno. Justifica por qué el tipado dinámico de Python facilita o complica esta tarea frente a un lenguaje de tipado estático.
*Evalúa:* manejo de tipos de datos básicos y pensamiento comparativo.



**7. Diseñando un experimento de muestreo**
*Situación:* Un investigador necesita generar 30 puntos de temperatura igualmente espaciados entre 10°C y 40°C para una simulación. Resuelve el problema con `np.linspace()` y con `np.arange()`, y argumenta cuál es más apropiada quirúrgicamente para este caso (conoces el número de puntos, no el paso).
*Evalúa:* diferenciación funcional entre `linspace` y `arange` (competencia clave del Parcial 1).



**8. Biblioteca reutilizable de progresiones**
*Situación:* Tu profesor necesita generar progresiones aritméticas para distintos parciales sin reescribir código cada vez. Diseña la función `progresion(inicio, fin, n_terminos)` pedida en el cuaderno, pruébala con tres casos, y documenta su comportamiento con *docstring*.
*Evalúa:* abstracción funcional y buenas prácticas de documentación.



**9. Python en el mundo real: elige tu dominio**
*Situación:* Elige uno de los cuatro sectores mencionados (salud, finanzas, analítica de datos, IA/ML), investiga un caso real de uso de Python en ese sector y las bibliotecas involucradas (NumPy, Pandas, Scikit-learn, TensorFlow), y presenta un mini-caso de estudio con un ejemplo de código simplificado.
*Evalúa:* conexión entre teoría y aplicación profesional, investigación guiada.



**10. Problema integrador: simulación completa**
*Situación (cierra el ABP):* Como preparación final para el Parcial 1, se te entrega un escenario: "una fábrica mide la producción diaria durante N días, conoce el primer y último valor". Debes: (a) declarar variables con tipos correctos, (b) importar NumPy, (c) construir la progresión con `linspace` **y** con `arange` según los datos disponibles en cada subcaso, (d) graficar el resultado con `matplotlib` usando cualquiera de los tres paradigmas vistos, y (e) documentar todo el proceso en Markdown.
*Evalúa:* integración de **todos** los objetivos de aprendizaje del capítulo en un solo producto evaluable — el objetivo explícito que el propio cuaderno declara para el Parcial 1.
