## Pregunta orientadora

¿Cómo podemos organizar información sobre números naturales para consultar y analizar sus propiedades mediante Python?

## 1. Introducción: organizar información matemática

En ciencia de datos no basta con almacenar números: también necesitamos asociarles información que permita interpretarlos. Para estudiar números naturales podemos registrar sus divisores, la cantidad de divisores y si son primos.

Un **diccionario** de Python es una estructura de datos que asocia cada **clave** con un **valor**. Las claves permiten identificar la información que queremos recuperar.


## 2. Crear un diccionario

Vamos a crear una ficha matemática para el número 7. Recordemos que 7 es primo y sus divisores positivos son 1 y 7.


```python
numero_6 = {'numero':6, 
           'es_primo':True, 
           'divisores':[1,2,3,6], 
           'cantidad_divisores':4}
```

```python
numero_6
```

```python
# Diccionario con información sobre el número 7

numero_7 = {
    "numero": 7,
    "es_primo": True,
    "divisores": [1, 7],
    "cantidad_divisores": 2
}

print(numero_7)
```

### Interpretación

- `numero_6` es la variable que almacena el diccionario.
- Las llaves `{}` delimitan el diccionario.
- `"numero"`, `"es_primo"`, `"divisores"` y `"cantidad_divisores"` son las claves.
- `6`, `True`, `[1, 2, 3, 6]` y `4` son los valores asociados.

Los valores pueden ser de distintos tipos: enteros, booleanos, listas, cadenas de texto, entre otros.


```python
numero_6.items() # el método items() sirve para convertir el diccionario en una lista de tuplas, 
# donde las componentes de estos pares ordenados son las parejas clave valor
```

```python
# Consultar la estructura clave-valor de forma legible

for clave, valor in numero_6.items():
    print(f"Clave: {clave!r:24} | Valor: {valor}")
```

## 3. Añadir nuevas parejas clave-valor

Podemos ampliar la ficha cuando necesitamos registrar nuevas propiedades. La instrucción `diccionario[clave] = valor` crea una nueva pareja si la clave no existe; si ya existe, reemplaza su valor.


```python
# Agregar nuevas propiedades al número 7

numero_6["es_impar"] = False
numero_6["cuadrado"] = 6 ** 2
numero_6["cubo"] = 6 ** 3

print(numero_6)
```

```python
numero_6
```

### Actualizar un valor existente

El siguiente ejemplo modifica intencionalmente el cuadrado de 7 para mostrar cómo funciona la actualización. **El valor 50 es incorrecto matemáticamente**: el cuadrado de 7 es 49. Después corregiremos el dato.


```python
numero_6["cuadrado"] = 17
print("Cuadrado registrado:", numero_6["cuadrado"])
```

```python
# Corregir el dato para mantener la consistencia matemática
numero_6["cuadrado"] = 6 ** 2
print("Cuadrado corregido:", numero_6["cuadrado"])
```

## 4. Acceder a los valores mediante sus claves

Cuando conocemos la clave de una propiedad, podemos recuperar su valor escribiendo `diccionario[clave]`.


```python
numero_6['cubo']
```

```python
print("Número:", numero_7["numero"])
print("¿Es primo?", numero_7["es_primo"])
print("Divisores:", numero_7["divisores"])
print("Cantidad de divisores:", numero_7["cantidad_divisores"])
print("Cuadrado:", numero_7["cuadrado"])
print("Cubo:", numero_7["cubo"])
```

### Consultar con `get()`

El método `get()` permite consultar una clave sin provocar un `KeyError` si no existe. Podemos indicar un valor predeterminado.


```python
print(numero_6.get("cantidad_divisores"))
```

```python
print(numero_6.get("suma_divisores", "Dato no registrado"))
```

## 5. Aplicación a ciencia de datos: varios registros matemáticos

Cada diccionario puede representar una observación. La lista `datos` reúne varios registros, que después podemos recorrer para consultar propiedades o convertirlos en una tabla de Pandas.


```python
# Datos de tres números naturales

numero_5 = {
    "numero": 5,
    "es_primo": True,
    "cantidad_divisores": 2
}

numero_6 = {
    "numero": 6,
    "es_primo": False,
    "cantidad_divisores": 4
}

numero_9 = {
    "numero": 9,
    "es_primo": False,
    "cantidad_divisores": 3
}
```

```python
# Añadir el cuadrado a cada registro
numero_5["cuadrado"] = 5 ** 2
numero_6["cuadrado"] = 6 ** 2
numero_9["cuadrado"] = 9 ** 2
```

```python
numero_5
```

```python
# Colección de registros
datos = [numero_5, numero_6, numero_9]

for registro in datos:
    print(
        registro["numero"],
        registro["es_primo"],
        registro["cantidad_divisores"],
        registro["cuadrado"]
    )
```

### Convertir los registros en una tabla con Pandas

Este paso conecta las estructuras básicas de Python con una herramienta muy utilizada en ciencia de datos. No es indispensable para comprender los diccionarios, pero muestra cómo pueden aprovecharse después.


```python
import pandas as pd

df = pd.DataFrame(datos) # Una dataframe se puede formar a partir de una lista de diccionarios 
df
```

## 6. Reto: detectar cuadrados perfectos

**Pregunta:** ¿cómo podemos agregar a cada registro una clave llamada `"es_cuadrado_perfecto"` que indique si el número es un cuadrado perfecto?

Por ejemplo, \(9 = 3^2\), por lo que 9 es un cuadrado perfecto. En cambio, 5 y 6 no lo son.

Primero intenta resolverlo por tu cuenta. Luego ejecuta la solución.


```python
from math import isqrt, sqrt
isqrt(5), sqrt(5)
```

```python


for registro in datos:
    n = registro["numero"]
    registro["es_cuadrado_perfecto"] = isqrt(n) ** 2 == n

for registro in datos:
    print(registro["numero"], registro["es_cuadrado_perfecto"])
```

```python
# Actualizar la tabla para incluir la nueva propiedad

df = pd.DataFrame(datos)
df
```

## 8. Resumen

| Operación | Ejemplo |
|---|---|
| Crear un diccionario | `d = {"numero": 7, "es_primo": True}` |
| Añadir o actualizar una pareja | `d["cuadrado"] = 49` |
| Consultar por clave | `d["numero"]` |
| Consultar con valor predeterminado | `d.get("suma_divisores", 0)` |
| Recorrer las parejas | `for clave, valor in d.items(): ...` |

**Idea central:** los diccionarios permiten organizar propiedades matemáticas mediante claves descriptivas. Combinados con listas, ciclos y Pandas, ayudan a construir conjuntos de datos, verificar propiedades y preparar información para investigaciones en ciencia de datos.
