### BLOQUE 1: Gráficos de Líneas y Dispersión (60 minutos)

**Contexto:** En Métodos Numéricos I, resolverán ecuaciones no lineales como \(f(x) = 0\) (bisección, Newton-Raphson). Antes de aplicar el método, es vital graficar la función para ubicar visualmente las raíces.

*   **Teoría (15 min):**
    *   Sintaxis de `plt.plot(x, y)`.
    *   Personalización: `color`, `linestyle`, `marker`, `label`.
    *   Títulos y etiquetas: `plt.title()`, `plt.xlabel()`, `plt.ylabel()`, `plt.legend()`, `plt.grid()`.
    *   Sintaxis de `plt.scatter(x, y)`.

*   **Ejemplo Guiado Teoría de Números (15 min):**
    Graficar la función $f(x) = x^2 - x - 6$ (para encontrar sus raíces en $x = -2$ y $x = 3$).


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 5, 100)
y = x**2 - x - 6

plt.plot(x, y, color='blue', label='f(x) = x^2 - x - 6')
plt.axhline(0, color='red', linestyle='--') # Eje X (para ver raíces)
plt.title('Gráfico de la Función Cuadrática')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.xticks(np.arange(-4, 6, 1))
plt.legend()
plt.grid(True)
plt.show()
````
