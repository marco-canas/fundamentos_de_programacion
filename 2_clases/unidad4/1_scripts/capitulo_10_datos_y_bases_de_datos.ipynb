## Bloque 1 — Creación de archivos `.csv` desde DataFrames de Pandas (50 min)



**1.1 Contextualización**

Se plantea una situación concreta: se ha calculado, con un script en Python, la **función φ de Euler** para los enteros del 1 al 100. El resultado está en un DataFrame. La pregunta es: ¿cómo conservarlo para analizarlo después sin recalcular?

Se discute brevemente por qué CSV es útil: legible por humanos, compatible con Excel y con otras herramientas, adecuado para datos tabulares planos.



**1.2 Demostración guiada (20 min)**

Se construye el DataFrame de ejemplo:


````python
# función para el número de divisores de un número n, también llamada función tau(n)
def tau(n):
    count = 0
    for i in range(1, n + 1):
        if n % i == 0:
            count += 1
    return count    
````

````python
import numpy as np  
tau = np.vectorize(tau)  # Vectorizamos la función tau para que pueda aplicarse a arrays de NumPy
````

````python
import pandas as pd
n = np.arange(1, 21)
datos = {
    'n': n,
    'phi': tau(n),
    'es_primo': tau(n) == 2
}
df = pd.DataFrame(datos)
df
````

Exportación con parámetros comentados:



````python
df.to_csv('phi_hasta_20.csv', index=False, encoding='utf-8')
````


**Puntos clave a explicar:**
- `index=False`: evita que el índice del DataFrame se escriba como una columna adicional, lo que suele confundir al reimportar.
- `encoding='utf-8'`: relevante si en el futuro se añaden columnas con texto en español.
- `sep`: por defecto `,`; en contextos donde se use Excel en configuración regional colombiana, puede requerirse `sep=';'`.

## Bloque 2 — Creación de bases de datos SQLite con `sqlite3` (55 min)



**2.1 Motivación (10 min)**

Se retoma el ejemplo anterior: si en lugar de 20 números se tienen 100.000, y se quiere responder repetidamente preguntas como “¿cuántos primos hay entre 50.000 y 60.000?” o “¿cuál es el promedio de φ para n compuesto?”, releer todo el CSV en cada consulta es ineficiente. Una base de datos permite **indexar, filtrar y agregar** sin cargar todo en memoria.

SQLite se presenta como: biblioteca estándar de Python (`sqlite3`), sin instalación adicional, archivo único `.db`, ideal para aprender SQL sin administrar un servidor.



**2.2 Creación de la base de datos y la tabla (20 min)**

Demostración paso a paso:


````python
import sqlite3

conexion = sqlite3.connect('teoria_numeros.db')
cursor = conexion.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS funcion_phi (
        n INTEGER PRIMARY KEY,
        phi INTEGER NOT NULL,
        es_primo INTEGER NOT NULL
    )
''')
conexion.commit()
````

**Explicación de decisiones de diseño:**
- `n INTEGER PRIMARY KEY`: garantiza unicidad y acelera búsquedas por número.
- `es_primo INTEGER` en lugar de `BOOLEAN`: SQLite no tiene tipo booleano nativo; se usa 0/1 y se documenta la convención.
- `NOT NULL`: evita filas incompletas.



**2.3 Inserción de datos (15 min)**



````python
filas = [(n, phi, 1 if primo else 0) for n, phi, primo in
         zip(df['n'], df['phi'], df['es_primo'])]
# ojo, las filas son listas de tuplas, no un DataFrame. 
cursor.executemany(
    'INSERT OR REPLACE INTO funcion_phi (n, phi, es_primo) VALUES (?, ?, ?)',
    filas
)
conexion.commit()
````

Se discute `INSERT OR REPLACE` frente a `INSERT` simple: el primero permite reejecutar el script sin errores por clave duplicada.



**2.4 Consulta directa con SQL (10 min)**



````python
cursor.execute('SELECT n, phi FROM funcion_phi WHERE es_primo = 0')
for fila in cursor.fetchall():
    print(fila)
````

Los estudiantes ejecutan una consulta propia: los números `n` donde `phi(n) = n - 1` (que corresponde a primos, propiedad elemental).




#### Bloque 3 — Acceso a bases de datos SQL con Pandas (55 min)

**3.1 Del cursor a Pandas (10 min)**

Se explica que `pd.read_sql()` envuelve la conexión y devuelve un DataFrame directamente, lo que evita iterar con `fetchall()` y facilita el análisis posterior.



````python
import pandas as pd

df_primos = pd.read_sql(
    'SELECT n, phi FROM funcion_phi WHERE es_primo = 1 ORDER BY n',
    conexion
)
df_primos 
````


**3.2 Consultas parametrizadas y agregaciones (20 min)**

Se construyen consultas que responden preguntas de Teoría de Números con estructura de análisis descriptivo:



````python
# Promedio de phi según condición de primalidad
df_resumen = pd.read_sql('''
    SELECT
        es_primo,
        COUNT(*) AS cantidad,
        AVG(phi) AS phi_promedio,
        MIN(phi) AS phi_min,
        MAX(phi) AS phi_max
    FROM funcion_phi
    GROUP BY es_primo
''', conexion)
df_resumen  
````

Esta consulta produce una tabla que **ya es un insumo directo** para el Análisis Descriptivo de Datos que verán a continuación: medias, conteos, extremos por grupo.



**3.3 Escritura desde Pandas hacia SQL (15 min)**

Se muestra el flujo inverso: si se calcula una nueva tabla (por ejemplo, la **función τ(n)** de número de divisores) en Pandas, se puede persistir en la misma base de datos:


````python
def contar_divisores(n):
    return sum(1 for i in range(1, n + 1) if n % i == 0)

df_tau = pd.DataFrame({
    'n': list(range(1, 21)),
    'tau': [contar_divisores(n) for n in range(1, 21)]
})

df_tau.to_sql('funcion_tau', conexion, if_exists='replace', index=False)
````

Se explican las opciones de `if_exists`: `'fail'`, `'replace'`, `'append'`.


### Ejercicio 2 — Conteo de primos por intervalo a CSV

**Contexto matemático.** La función π(x) cuenta los primos ≤ x. Se desea una tabla que, para intervalos de tamaño 10 hasta 100, registre cuántos primos hay en cada intervalo.

**Tarea:**
1. Construya un DataFrame con columnas `intervalo` (texto, ej. `"1-10"`, `"11-20"`, …) y `cantidad_primos` para los intervalos [1,10], [11,20], …, [91,100].


````python
#**Tarea:**
#1. Construya un DataFrame con columnas `intervalo` (texto, ej. `"1-10"`, `"11-20"`, …) y
# `cantidad_primos` para los intervalos [1,10], [11,20], …, [91,100].

import pandas as pd
def tau(n):
    count = 0
    for i in range(1, n + 1):
        if n % i == 0:
            count += 1
    return count  

df_primos = pd.DataFrame({
    'intervalo': [f"{i}-{i+9}" for i in range(1, 100, 10)],
    'cantidad_primos': [sum(1 for n in range(i, i + 10) if tau(n) == 2) 
                        for i in range(1, 100, 10)]
})
df_primos 
````

# Solución completa del ejercicio 3  

import sqlite3

# ---------------------------------------------------------------
# Funciones auxiliares
# ---------------------------------------------------------------
def es_primo(n):
    """Devuelve True si n es primo, False en caso contrario."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


# ---------------------------------------------------------------
# P2.1 — Conexión y creación del archivo .db
# ---------------------------------------------------------------
conexion = sqlite3.connect('congruencias.db')
cursor = conexion.cursor()

# ---------------------------------------------------------------
# P2.2 — Creación de la tabla con restricciones
# ---------------------------------------------------------------
cursor.execute('''
    CREATE TABLE IF NOT EXISTS modulo7 (
        n         INTEGER PRIMARY KEY,
        residuo   INTEGER NOT NULL,
        es_primo  INTEGER NOT NULL
    )
''')
conexion.commit()

# ---------------------------------------------------------------
# P2.4 — Inserción masiva con executemany
# ---------------------------------------------------------------
# Construimos las filas: (n, n % 7, 1 si primo else 0)
filas = [
    (n, n % 7, 1 if es_primo(n) else 0)
    for n in range(1, 31)
]

cursor.executemany(
    'INSERT OR REPLACE INTO modulo7 (n, residuo, es_primo) VALUES (?, ?, ?)',
    filas
)
conexion.commit()

print(f"Se insertaron {len(filas)} filas en la tabla 'modulo7'.")

# ---------------------------------------------------------------
# P2.5 — Consulta directa con cursor
# ---------------------------------------------------------------
cursor.execute('''
    SELECT n, residuo, es_primo
    FROM modulo7
    WHERE residuo = 3 AND es_primo = 1
    ORDER BY n
''')
resultados = cursor.fetchall()

print("\nNúmeros primos con residuo 3 (mód 7):")
print(" n  | residuo | es_primo")
print("----+---------+----------")
for fila in resultados:
    print(f" {fila[0]:>2} |    {fila[1]}    |    {fila[2]}")

# ---------------------------------------------------------------
# P2.6 — Cierre de la conexión
# ---------------------------------------------------------------
cursor.close()
conexion.close()
print("\nConexión cerrada. Base de datos 'congruencias.db' creada.")
