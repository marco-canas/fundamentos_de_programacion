```python
import numpy as np 
import pandas as pd 

N = 100
datos = []
for n in range(1,N+1):
    divisores = [d for d in range(1, n+1) if n%d == 0]
    es_primo = len(divisores) == 2
    num_divisores = len(divisores)
    sum_divisores = sum(divisores)
    datos.append({
        'n':n, 
        'divisores': divisores, 
        'es_primo': es_primo, 
        'num_divisores': num_divisores, 
        'sum_divisores':sum_divisores
    })
```

```python
datos
```

```python
df = pd.DataFrame(datos)
```

```python
df.to_csv('propiedades_numeros.csv', index = False, encoding = 'utf-8' )
```
