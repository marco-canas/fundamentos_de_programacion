```python
import numpy as np 
import pandas as pd 
```

```python
datos = [] 

```

```python
for n in range(5_000, 10_001):
    divisores_propios = [divisor for divisor in range(1, n) if n%divisor == 0]
    es_primo = len(divisores_propios) == 1
    suma_divisores_propios = sum(divisores_propios)
    es_perfecto = suma_divisores_propios == n
    es_abundante = suma_divisores_propios > n
    es_deficiente = suma_divisores_propios < n
    datos.append({
        'n': n,
        'num_divisores_propios': len(divisores_propios),
        'es_primo': es_primo, 
        'suma_divisores_propios':suma_divisores_propios,
        'es_perfecto': es_perfecto, 
        'es_abundante': es_abundante, 
        'es_deficiente':es_deficiente
    })
    
```

```python
datos
```

```python
df = pd.DataFrame(datos)
```

```python
import sqlite3
```

```python
df['es_primo'].astype(int)
df['es_perfecto'].astype(int)
df['es_abundante'].astype(int)
df['es_deficiente'].astype(int)
```

```python
base_datos = sqlite3.connect('aritmetica.db')
cursor = base_datos.cursor() 
```

```python
# 2. Creación de tabla
cursor.execute('''
    CREATE TABLE IF NOT EXISTS tabla1 (
        n INTEGER PRIMARY KEY, 
        num_divisores_propios INTEGER,
        es_primo INTEGER,
        suma_divisores_propios INTEGER, 
        es_perfecto INTEGER, 
        es_abundante INTEGER, 
        es_deficiente INTEGER
    )
''')
```

```python
# inserción masiva de los datos 

datos_insercion = df.values.tolist()  

cursor.executemany('''
    INSERT OR REPLACE INTO tabla1 (n, num_divisores_propios, es_primo, 
    suma_divisores_propios, es_perfecto, es_abundante, es_deficiente)
    VALUES (?, ?, ?, ?, ?, ?, ?)
''', datos_insercion)
```

```python
base_datos.commit()
base_datos.close()
```
