## 8. Consolidado, exportación y seguimiento histórico### 8.1 Guardar el registro de esta sesiónExporta lo registrado a archivos CSV acumulativos. Cada vez que corras esta celda al final de una asesoría, la sesión se añade al histórico general (no se sobreescribe), permitiendo construir el historial del estudiante a lo largo del curso.

```python
import osCARPETA_HISTORIAL = "historial_asesorias"os.makedirs(CARPETA_HISTORIAL, exist_ok=True)def guardar_sesion():    resumen_sesion = pd.DataFrame([{        **sesion,        **{f"plan_{k}": (", ".join(v) if isinstance(v, list) else v) for k, v in plan_accion.items()},        **{f"cierre_{k}": v for k, v in cierre_metacognitivo.items()},    }])    rutas = {        "sesiones.csv": resumen_sesion,        "dificultades.csv": registro_dificultades,        "estrategias.csv": registro_estrategias,    }    for nombre_archivo, df in rutas.items():        ruta = os.path.join(CARPETA_HISTORIAL, nombre_archivo)        if os.path.exists(ruta):            df_previo = pd.read_csv(ruta)            df_final = pd.concat([df_previo, df], ignore_index=True)        else:            df_final = df        df_final.to_csv(ruta, index=False)    print(f"Sesión guardada. Archivos en ./{CARPETA_HISTORIAL}/")# guardar_sesion()   # descomentar y ejecutar al terminar la asesoría
```

### 8.2 Seguimiento histórico de un estudianteConsulta rápidamente el historial acumulado de un estudiante en particular, para preparar la siguiente asesoría sabiendo qué ya se intentó.

```python
def historial_estudiante(nombre_estudiante):    ruta_dif = os.path.join(CARPETA_HISTORIAL, "dificultades.csv")    ruta_est = os.path.join(CARPETA_HISTORIAL, "estrategias.csv")    if not os.path.exists(ruta_dif):        print("Aún no hay historial guardado.")        return None, None    dif = pd.read_csv(ruta_dif)    est = pd.read_csv(ruta_est) if os.path.exists(ruta_est) else pd.DataFrame()    dif_estudiante = dif[dif["estudiante"] == nombre_estudiante]    est_estudiante = est[est["estudiante"] == nombre_estudiante] if not est.empty else est    print(f"--- Dificultades históricas de {nombre_estudiante} ---")    display(dif_estudiante)    print(f"--- Estrategias aplicadas con {nombre_estudiante} ---")    display(est_estudiante)    if not dif_estudiante.empty:        print("\nTemas recurrentes:")        display(dif_estudiante["tema"].value_counts())    return dif_estudiante, est_estudiante# historial_estudiante("Nombre del estudiante")
```
